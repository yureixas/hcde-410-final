# Background

The internet has made information increasingly accessible across the world, but access to information online is not always equal across languages. Much of the information available online may have more representation in certain languages, which can create barriers for people who may need access to information in other languages. Wikipedia is an example of this issue because it is an online encyclopedia that can show content in different languages. However, the existence of the content in those languages does not mean that it contains the same range of information as other languages. According to the [List of Wikipedia statistics](https://meta.wikimedia.org/wiki/List_of_Wikipedias), the number of articles varies considerably between different languages. One important consideration is that the number of articles does not indicate how useful, accurate, or representative Wikipedia is for a certain language. It only shows th the amount of content available.


## Related Work

Johnson and Lescak argue that across different language editions of Wikipedia, they do not consistently show the same content. They talk about three factors that affect Wikipedia data across languages: local context, community and governance, and technology. This helps inform the direction of how I should interpret the data in my project, as a low article count may reflect different norms that an editing community has and may not be due to a lack of information available to speakers. 

Miquel-Ribé and Laniado built the Wikipedia Diversity Observatory to help communities identify content gaps on Wikipedia, including cultural gaps. Their work shows that gaps in Wikipedia are a real concern that Wikipedia has identified, and also gives me a point of comparison for my data analysis. I will draw on their published findings and repository to interpret my results.

Ritchie et al. created LinguaMeta, a metadata resource for thousands of languages that includes estimated speaker populations by their location and discusses their methodology for recording estimates of the number of speakers per language. I will use LinguaMeta as my source for comparing speaker counts to Wikipedia content.

These sources will help me in my data analysis by providing different contexts and reasons that might affect my data analysis. My checks will look for any biases in the data, such as skews in article counts from bot-generated content.
